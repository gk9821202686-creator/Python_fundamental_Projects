# Guess The Number Game

```python
import random
difficulty_level = int(input("Game modes \n 1 = Easy[1-10] \n 2 = Medium[1-50] \n 3 = hard[1-100]\n Select"))
match difficulty_level:
    case 1:
        r = random.randint(1, 10)
        print("difficulty level set to Easy")
        i=1
        while i<10:
            g = int(input("guess the number"))
            if g > r:
                print("lower")
            elif g < r:
                print("higher")
            else: 
                print(f"YOU WON \n In {i} Guess ")
                break
            i = i + 1
    case 2:
        r = random.randint(1, 50)
        print("difficulty level set to Medium")
        i=1
        while i<50:
            g = int(input("guess the number"))
            if g > r:
                print("lower")
            elif g < r:
                print("higher")
            else: 
                print(f"YOU WON \n In {i} Guess ")
                break
            i = i + 1
    case 3:
        r = random.randint(1, 100)
        print("difficulty level set to hard")
        i=1
        while i<100:
            g = int(input("guess the number"))
            if g > r:
                print("lower")
            elif g < r:
                print("higher")
            else: 
                print(f"YOU WON \n In {i} Guess ")
                break
            i = i + 1
    case _:
        print("Oops something went wrong")
```
